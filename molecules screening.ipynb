```python
import pandas as pd
df = pd.read_csv(r"C:\Users\Rotana Store\Desktop\cheminformatics\data_samples\drug_discovery_molecules.csv")
def screen_molecule (molecule):
  violations = 0
  violations_list = []
  if molecule['MW'] <= 500 :
     pass
  else : 
   violations +=1 
   violations_list.append('MW')
  if molecule['LogP'] <= 5 :
     pass
  else : 
   violations +=1 
   violations_list.append('LogP')
  if molecule['HBD'] <= 5 :
     pass
  else : 
   violations +=1 
   violations_list.append('HBD')
  if molecule['HBA'] <= 10 :
     pass
  else :
   violations +=1 
   violations_list.append('HBA')
  if violations == 0 :
      status = 'pass'
  elif violations == 4 :
      status = 'fail'
  else :
      status ='review'
  return status, violations , violations_list

molecules_results = []
for index, molecule in df.iterrows() :
  status , violations , violations_list = screen_molecule(molecule)
  molecules_result = {'name': molecule['name'], 'status': status, 'violations': violations, 'violations_list': violations_list}
  molecules_results.append(molecules_result)

df['status'] = [result['status'] for result in molecules_results]
df['violations'] = [result['violations'] for result in molecules_results]
df['violations_list'] = [result['violations_list'] for result in molecules_results]
print(df['status'].value_counts())

MW_violations = 0
MW_violations_molecules = []
LogP_violations = 0
logP_violations_molecules = []
HBD_violations = 0
HBD_violations_molecules = []
HBA_violations = 0
HBA_violations_molecules = []
for index, row in df.iterrows() :
   if 'MW' in row['violations_list'] : 
      MW_violations +=1
      MW_violations_molecules.append(row['name'])
   if 'LogP' in row['violations_list'] : 
      LogP_violations +=1
      logP_violations_molecules.append(row['name'])
   if 'HBD' in row['violations_list'] :
      HBD_violations +=1
      HBD_violations_molecules.append(row['name'])
   if 'HBA' in row['violations_list'] :
      HBA_violations +=1
      HBA_violations_molecules.append(row['name'])

print('MW violations:', MW_violations , 'LogP violations:', LogP_violations , 'HBD violations:', HBD_violations, 'HBA violations:', HBA_violations)
   

print(len(df[df['status'] == 'pass']))
print(df.to_string())

df.to_csv('screening_results.csv', index=False)
```
